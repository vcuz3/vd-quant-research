#### Exercise 4.1

```python
""" A strategy wins 44 percent of the time with average win 1.9 and average
loss 1.0 units, costs 0.05 per trade. Compute its expectancy, then the expected profit of
300 trades risking 1 percent of a 50,000 account per trade, ignoring compounding. """

win_rate = 0.44
loss_rate = 1 - win_rate
outcome_win = 1.9
outcome_loss = -1
cost = 0.05
risk_pct = 0.01
initial_balance = 50000
expectancy = win_rate * (outcome_win - cost) + loss_rate * (outcome_loss - cost)
expected_profit = (300 * risk_pct * initial_balance) * expectancy
print("expectancy = " + str(np.round(expectancy, 2)))
print("expected_profit = " + str(np.round(expected_profit, 2)))
```

#### Exercise 5.1

```python
""" A strategy's monthly returns have arithmetic mean 1.2 percent and volatility
6 percent. Estimate its annual geometric growth rate, and its terminal multiple over 10
years. """
annual_return = 0.012 * 12
annual_vol = 0.06 * np.sqrt(12)
est_geo_mean = annual_return - ((annual_vol ** 2) / 2)
ten_yr_multiple = (1 + est_geo_mean) ** 10
print(ten_yr_multiple)
```

#### Exercise 5.2

```python
""" Your account is down 18 percent. Compute the recovery gain required. Now
compute it if you first average down and the drawdown reaches 34 percent. Write one
sentence to your future self about what this implies for adding to losers. """
print("recovery gain given 18% drawdown: ", str(np.round(0.18 / (1 - 0.18) * 100, 2)))
print("recovery gain given 34% drawdown: ", str(np.round(0.34 / (1 - 0.34) * 100, 2)))
# a ~1.9 increase in drawdown here required making 50% to catch up back to even. avoiding
# deep drawdown is absolutely essential in compounding terminal wealth.
```
