```python
""" Rework Example 4.1 with a researcher whose process is hypothesis-first,
raising the prior to 5 percent, and whose testing discipline lowers the false positive rate to
2 percent. What is a great backtest worth now? Reflect on which change did more work. """

""" Exercise 4.2 with the fpr only leg """
fpr = 0.02
edge_prob = 0.01
great_backtest_given_edge = 0.6
prob_edge_given_great_backtest = (great_backtest_given_edge * edge_prob) / (
    great_backtest_given_edge * edge_prob + (1 - edge_prob) * fpr)
print("probability of edge given a great backtest after reducing fpr and keeping p(edge) constant: " + str(np.round(prob_edge_given_great_backtest, 2)))
```

```python
# From previous working, raising the probability of edge alone raises the P(edge | great backtest) to 0.39, a 0.28 increase.
# Here, keeping edge at 0.01, reducing fpr raises the P(edge | great backtest) to 0.23, a 0.12 increase in prob. Both together
# increases prob by 0.5, so the prior did more and the effects compounded.
```
