```python
# 1. Define edge in one sentence containing all four load-bearing clauses.
""" An edge is a repeatable process whose expected profit is 
positive after all costs, with a variance profile you can survive 
long enough for the expectation to assert itself. 
The 4 load-bearing clauses:
1) Repeatable: excludes one-off windfalls and information you cannot get again.
2) Expected: excludes any single outcome.
3) After all costs: excludes most published strategies, which die the moment spreads, fees, and slippage are applied.
4) Survival: excludes strategies whose drawdowns bankrupt you before the long run arrives. A positive expectation you
cannot survive is not an edge. 
"""
```

```python
# 2. A strategy wins 38 percent with W/L of 2.4 and costs of 0.08 per trade. Expectancy?
expectancy = 0.38 * (2.4 - 0.08) + (1 - 0.38) * (-1 - 0.08)
print(np.round(expectancy, 2))
```

```python
# 4. Name the five sources of real edge and give a concrete example of a structural one.
""" Five sources of real edge:
1. risk premium: payment for bearing risk others want to shed: equity risk, volatility risk, liquidity risk. 
These are the most durable edges because they are compensation, not exploitation. They also come with the
drawdowns you are being paid to hold.
2. Behavioural patterns: systematic mistakes by other particiapnts, such as overreaction, underreaction,
and anchoring, leave statistical traces. These edges are real but decay as they become known and crowded.
3. Structural and institutional constraints: index funds must rebalance on schedule. Some funds cannot
hold certain assets. Options dealers must hedge. Participants forced to trade for non-price reasons pay 
whoever takes the other side.
4. Information and speed: knowing something first, or acting on it faster. Almost entirely closed to individual
traders at short horizons, which has an important implication: your hunting ground is horizons and niches
where the fast money does not bother to compete.
5. Service provision: supplying liquidity and earning the spread. This is a business with real inventory risk.

Examples of structural edges:
1. An US company that needs to make an international order to another company in AUD needs to convert money
regardless of the current price.
2. A hedge fund with an open position that is in deeper drawdown than their risk boundary can endure, so they
must reduce exposure and exit the position, whether price is favourable or not.
"""
```

```python
# 6. State the four behavioral signatures and the trade-log statistic that detects each
""" Four behavioural signatures:
1. Overtrading: trading for stimulation or to make something back, far beyond any signal. Statistical signature: 
trade frequency is uncorrelated with opportunity measures, and cost drag as a large share of gross losses.
2. Disposition effect: selling winners early to lock in being right, holding losers to avoid admitting being
wrong. Signature: average holding time of losers exceed that of winners, and the win/loss payoff ratio sits well
below 1 while win rate looks respectable.
3. Revenge trading: size and frequency spike immediately after losses. Signature: position size positively 
correlated with recent drawdown, exactly backwards from any typical risk process does. Variance explodes precisely
when capital is most depleted.
4. Sample-size inaccuracy: full confidence after 15 wining trades, system abandonment after 8 losing ones. 
Signature: strategy switching at a frequency that no statistical evidence could justify.
"""
```

```python
# 7. What are the three defining properties of a systematic approach?
""" 3 defining properties of a systematic approach:
1. The rule is explicit: given these inputs, take these actions, with no residual judgement at execution time.
2. The rule is testable: you can ask precisely what that explicit rule would have done across history and what
it is expected to do across plausible futures.
3. The rule is stable under pressure: it executes the same way after 3 losses as after 3 wins, because it does
not know or care about what just happened.
"""
```

```python
# 8. Explain the evaluation asymmetry between a track record and a rule.
""" A track record from a discretionary trader is a set of decisions/outcomes that is difficult to falsify
or evaluate historically. This is  because the justification for a position is not written explicitly, so 
one can't confirm its performance across time, and replication failures can be blamed to external factors
(such as psychology) that is unfalsifiable. Moreover, a track record stemming from a small sample give no
predictive value in its future performance or across different regimes and indistinguishable from luck.
A rule, on the other hand, is directly testable and can be evaluated historically and across different 
regimes, given its explicit nature.
"""
```

```python
# 9. State Bayes' rule and identify which term validation attacks and which term hypothesis-first research attacks.
""" Bayes' rule is P (edge | evidence) = P (evidence | edge) * P(edge) / P(evidence)
P(evidence) is broken down to (P evidence | edge) * P(edge) + P(evidence | no edge) * P(no edge)
In words, the probability that a strategy has a genuine edge given a great backtest is the probability 
of a great backtest arising from a genuine edge, divided by the probability of a great backtest arising 
from either a genuine edge or no edge.
The validation attacks the term P(evidence | no edge), the false positive rate.
Hypothesis-first research attacks the probability of finding genuine edge.
"""
```

```python
# 15. Why does oversizing destroy a profitable strategy? Use the scaling of mean versus variance
""" Oversizing destroys a profitable strategy because of the asymetry between the scaling of mean
vs variance. Doubling the size on a position doubles the mean, but quadruples the variance (because
variance grows at size squared). This creates a larger volatility drag, and typical losses in a 
strategy now need unrealistic gains to recover and make profits.
"""
```

```python
# 17. Give one structural fact each about futures, crypto, and prediction markets that changes strategy design.
""" 
Futures: contracts expire every quarter, so a backtest using continuous futures data depends on how contracts
are stitched together. the roll method can materially change measured returns, so the backtest should model
contract rolls explicitly.
Crypto: trades continuous without a daily close. Strategies can face sharp, volatile moves at any hour, so risk
controls and position sizing needs to account for this.
Prediction markets: contracts resolve to $0 or $1 based on a defined event, and the outcome can sometimes be
disputed. Strategy design should account for resolution rules and dispute risk.
"""
```
