```python
# 1. Name the five species of financial data and the characteristic failure mode of each
""" The five species of financial data and characteristic failure mode of each are:
1) prices and volumes - adjustment errors, stale prints, bad ticks, bar-construction ambiguity.
2) fundamentals. earnings, balance sheets, filings - reporting lags misapplied, restatements, fiscal-calendar misalignment.
3) macro releases. GDP, inflation, employment, rates - revisions, publication lags, the gap between the period describe 
    and the moment is known
4) alternative data. anything else that might carry a signal: positioning reports, flows, sentiment, on-chain activity - 
    short histories, changing collection methodology, survivorship in the vendor's own product line
5) derived data. returns, spreads, factors, signals: everyting computer from the layers above - failure mode is inheritance,
    derived data is exactly as trustworthy as its worst input, and its computation adds fresh opportunities for error on top.
"""
```

```python
# 2. State the three questions asked of any dataset before use.
""" 
1) Where did this come from? Not which website: which venue, through which chain of intermediaries, retrieved when, under
    what terms. Provenance determmines which failure modes are live.
2) What was done to it? Adjustment, resampling, filtering, filling, revision: every dataset you will ever download has 
    been processed, and the processing embeds choices someone else made that may be is irrelevant, or worse, materially
    impact how you view the data.
3) What is missing from it? Delisted companies, gapped days, releases that were later revised, trades that were filtered.
    Missing data is difficult to tell given that there's no warning.
"""
```

```python
# 3. Walk the supply chain from venue to you and name one alteration at each hop.
"""
Data supply chain starts from the exchanges/primary venues -> consolidators and feeds -> vendors and free APIs -> user
At the consolidators stage, feeds are aggregated across venues into unified streams, and along the way the consolidators
    decide which trades count toward official prices, how out-of-sequence reports are handled, and what the canonical
    daily close of a fragmented market even means.
Below this, the vendors, from institutional platforms to free APIs, package the consolidated feeds into products:
    resampling it into bars, applying adjustment policies, mapping symbols across their own naming schemes, and 
    backfilling corrections into history, often silently.
"""
```

```python
# 4. Define point-in-time versus current-state, and state which one you can usually download
""" 
Point-in-time refers to the data that was observable at the given timestamp: what the GDP print said
    that morning, who was in the index that day, which companies existed and traded.
Current-state data refers to the data that is observable today, after all revisions and edits: today's
    membership, today's revised GDP figures, today's adjusted prices, today's symbol mappings.
Nearly everything that is freely downloadable is current-state, because it is what almost all users want;
    what already happened. 
"""
```

```python
# 6. Why is an adjusted series not stable over time, and what practice does that justify?
""" 
An adjusted series is not stable over time because every adjustment alters all the historical data; 
    e.g. a dividend, yield, futures roll, all rescale history. Even though an adjusted series
    reflect true investor experience.
This in turn justifies the practice of using the raw series instead if the research purpose is 
    regarding the prices people actually saw and transacted at. e.g. round-number behaviour, option
    strike interactions, price-level liquidity effects, historical positioning sizing in shares.
"""
```

```python
# 7. Name the three futures stitching methods and the question each answers correctly
"""
1) Unadjusted concatenation: preserves actual price levels and corrupts returns at every seam. It is
    only correct when levels themselves are the object of study.
2) Back-adjustment: shifts each earlier contract additively so the seams close. Price differences and
    point-based profit calculates are preserved, but distant history can be driven near or below zero,
    and percentage returns computed on those shifted levels are distorted.
3) Ratio adjustment: scales each earlier contract multiplicatively so the seams close in return space.
    Percentage returns are preserved throughout, levels become abstract index values.
Strategies reasoning in percentage returns (which is the majority) should use ratio-adjusted series.
"""
```

```python
# 8. Explain why survivorship bias overstates returns and understates risk simultaneously
""" Survivorship bias overstates returns by concentrating on the market where the losers are all
    hidden and deaths cluster, so the survivors-only series is both higher and smoother in returns,
    whilst also eliminating all actual drawdowns and overstating every risk-adjusted statistic
    simultaneously.
"""
```

```python
# 9. Why is a value strategy especially exposed to the graveyard?
"""
Value strategy specifically bets on the fact instruments which deviate from a fair value will
revert back to the mean. But stocks that fell far below the 'mean' and never recovered are removed
from the index and invisible to current-state. Only the survivors, stocks that fell and recovered 
are back-tested upon, and then the lesson is learnt brutally once traded live. 
"""
```

```python
# 18. Explain the asynchronous-close artifact and the test that dissolves it.
""" The asynchronous-close artifact is when two timeframe bars have a different close window but
overlapping body. A naive researcher assumes that the close is aligned to the same time. So the 
return calculated from one bar's close discloses information that the researcher thinks was not
available at the time, but was simply an artifact of the two closes having different definition.
The test that dissolves it alignment under the same UTC window, or lag the predictor to what was
actually knowable before the target market's open.
"""
```
